```python
total_sessions = len(df)

purchase_sessions = df["Revenue"].sum()

conversion_rate = purchase_sessions / total_sessions

print("总Session数:", total_sessions)
print("购买Session数:", purchase_sessions)
print("Session转化率:", round(conversion_rate * 100, 2), "%")
```
